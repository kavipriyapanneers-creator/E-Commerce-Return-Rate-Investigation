```python
print("=" * 50)
print("PHASE 4: OUTLIER DETECTION")
print("=" * 50)

import pandas as pd

numeric_columns = [
    "customer_age",
    "product_price",
    "discount_percent",
    "product_rating",
    "past_purchase_count",
    "past_return_rate",
    "delivery_delay_days",
    "session_length_minutes",
    "num_product_views"
]

outlier_results = []

for col in numeric_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_results.append([
        col, round(lower, 2),
        round(upper, 2), count
    ])

outlier_df = pd.DataFrame(
    outlier_results,
    columns=["Column", "Lower Limit", "Upper Limit", "Outlier Count"]
)

display(outlier_df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.bar(
    outlier_df["Column"],
    outlier_df["Outlier Count"]
)

plt.title("Outlier Detection using IQR Method")
plt.xlabel("Numerical Features")
plt.ylabel("Number of Outliers")

plt.xticks(rotation=60)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 7))

df[numeric_columns].boxplot()

plt.title("Outlier Distribution Across Numerical Features")
plt.ylabel("Values")
plt.xticks(rotation=60)

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import ttest_ind

returned = df[df["returned"] == 1]["delivery_delay_days"]
not_returned = df[df["returned"] == 0]["delivery_delay_days"]

t_stat, p_value = ttest_ind(
    returned,
    not_returned,
    equal_var=False
)

print("DELIVERY DELAY HYPOTHESIS TEST")
print("--------------------------------")

print("Returned Orders Mean:", returned.mean())
print("Non-Returned Orders Mean:", not_returned.mean())

print("T-Statistic:", t_stat)
print("P-Value:", p_value)

if p_value < 0.05:
    print("Result: Significant difference found.")
else:
    print("Result: No significant difference found.")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))

plt.boxplot(
    [not_returned, returned],
    tick_labels=["Not Returned", "Returned"]
)

plt.title("Delivery Delay vs Return Status")
plt.xlabel("Order Status")
plt.ylabel("Delivery Delay Days")

plt.show()
```

```python
from scipy.stats import ttest_ind

returned_discount = df[df["returned"] == 1]["discount_percent"]
not_returned_discount = df[df["returned"] == 0]["discount_percent"]

t_stat, p_value = ttest_ind(
    returned_discount,
    not_returned_discount,
    equal_var=False
)

print("DISCOUNT HYPOTHESIS TEST")

print("Returned Mean:", returned_discount.mean())
print("Not Returned Mean:", not_returned_discount.mean())

print("T-Statistic:", t_stat)
print("P-Value:", p_value)

if p_value < 0.05:
    print("Significant difference found.")
else:
    print("No significant difference found.")
```

```python
returned_rate = df[df["returned"] == 1]["past_return_rate"]
not_returned_rate = df[df["returned"] == 0]["past_return_rate"]

t_stat, p_value = ttest_ind(
    returned_rate,
    not_returned_rate,
    equal_var=False
)

print("PAST RETURN RATE HYPOTHESIS TEST")

print("Returned Mean:", returned_rate.mean())
print("Not Returned Mean:", not_returned_rate.mean())

print("T-Statistic:", t_stat)
print("P-Value:", p_value)

if p_value < 0.05:
    print("Significant difference found.")
else:
    print("No significant difference found.")
```

```python
print("=" * 50)
print("PHASE 6: RETURN RISK SEGMENTATION")
print("=" * 50)

# Create return risk segments
def segment_customer(rate):
    if rate < 0.15:
        return "Low Return Risk"
    elif rate < 0.30:
        return "Medium Return Risk"
    else:
        return "High Return Risk"

df["return_segment"] = df["past_return_rate"].apply(segment_customer)

# Display segment details
print("\nRETURN SEGMENT DETAILS")

segment_summary = df.groupby("return_segment").agg(
    total_orders=("returned", "count"),
    actual_returns=("returned", "sum"),
    actual_return_rate=("returned", "mean")
).reset_index()

segment_summary["actual_return_rate"] *= 100

display(segment_summary)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Graph 1: Number of orders in each segment
plt.figure(figsize=(8, 5))

sns.barplot(
    data=segment_summary,
    x="return_segment",
    y="total_orders"
)

plt.title("Order Count by Return Risk Segment")
plt.xlabel("Return Risk Segment")
plt.ylabel("Total Orders")
plt.xticks(rotation=15)
plt.show()


# Graph 2: Actual Return Rate
plt.figure(figsize=(8, 5))

sns.barplot(
    data=segment_summary,
    x="return_segment",
    y="actual_return_rate"
)

plt.title("Actual Return Rate by Risk Segment")
plt.xlabel("Return Risk Segment")
plt.ylabel("Actual Return Rate (%)")
plt.xticks(rotation=15)
plt.show()
```
